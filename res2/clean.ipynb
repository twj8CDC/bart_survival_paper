```python
fig,ax = plt.subplots(nrows=1, ncols=2, figsize=(20,8))
lw = "post"
ttt = [0, 90, 180, 270,360, 365]
lbl = [str(i) for i in ttt]

ax[0].step(ttt, [1]+psv0_m[:4]+[psv0_m[4]], where = lw, color="black", alpha=.8, linestyle="--", label="p-BART")
ax[1].step(ttt, [1]+psv1_m[:4]+[psv1_m[4]], where = lw, color="black", alpha=.4, linestyle="--", label="p-BART") 

ax[0].step(ttt, [1]+r_sv0m[:4]+[r_sv0m[4]], where = lw, color="black", linewidth=3, alpha=.8, linestyle="dotted", label= "r-BART")
ax[1].step(ttt, [1]+r_sv1m[:4]+[r_sv1m[4]], where = lw, color="black", linewidth=3, alpha=.4, linestyle="dotted", label= "r-BART")

ax[0].step(ttt, [1]+cph_r0[:4]+[cph_r0[4]], where = lw, color="black", alpha=0.8, linewidth = 4, linestyle="--", label="CPHM")
ax[1].step(ttt, [1]+cph_r1[:4]+[cph_r1[4]], where = lw, color= "black", alpha=0.4, linewidth=4, linestyle="--", label="CPHM")

plt.tight_layout()
ax[0].set_xticks(ttt[:5], lbl[:5])
ax[1].set_xticks(ttt[:5], lbl[:5])
ax[0].set_xlabel("Time (days)")
ax[1].set_xlabel("Time (days)")
ax[0].set_ylabel("S(t,x)")
ax[0].set_xlim(-0.01,370)
ax[1].set_xlim(-0.01,370)
ax[0].set_ylim(0.6,1.02)
ax[1].set_ylim(0.6,1.02)
ax[0].legend()
ax[1].legend()
```
